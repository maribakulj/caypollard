# 00 — Project overview

This notebook is the executable front door to **caypollard**. It states the research question, defines the experiment matrix, and checks that the small reusable utilities imported from `src/` behave as expected.

> **Primary question:** Can knowledge-graph structure improve retrieval of iconographically related cultural-heritage images beyond what visual embeddings alone recover?


## Hypotheses

- **H1:** graph representations contain complementary information to visual representations.
- **H2:** graph-aware retrieval improves visually dissimilar but iconographically related *hard positives*.
- **H3:** graph-aware retrieval suppresses visually similar but iconographically unrelated *hard negatives*.
- **H4:** part of the gain survives cross-collection transfer.
- **H5:** a learned joint model must beat transparent late fusion/reranking to justify its complexity.


## Planned experimental sequence

1. Iconclass corpus audit and hierarchy.
2. Visual baselines: DINOv2, CLIP/OpenCLIP, SigLIP.
3. KG baselines: Node2Vec, RDF2Vec, ComplEx, RotatE.
4. Frozen hard-positive / hard-negative benchmark.
5. Weighted late fusion and graph reranking.
6. Learned shared projection if simple fusion demonstrates complementary signal.
7. Emblematica historical case study.
8. Rijksmuseum cross-collection transfer.
9. WJoconde comparison and expert evaluation.

## Transparent fusion sanity check

Before any learned multimodal model exists, the project uses weighted late fusion as a mandatory baseline. The example below is deliberately tiny: it demonstrates the API, not a scientific result.


```python
visual = np.array([0.90, 0.75, 0.30])
graph = np.array([0.20, 0.80, 0.95])

for alpha in (0.0, 0.25, 0.5, 0.75, 1.0):
    print(alpha, late_fusion(visual, graph, alpha=alpha))
```

## Evaluation philosophy

Aggregate retrieval metrics are necessary but insufficient. Every major experiment should report:

- overall retrieval performance;
- hard-positive performance;
- hard-negative error rate;
- performance under group-aware splits;
- ranking changes caused by graph information;
- representative failures;
- confidence intervals / seed variance where appropriate.
